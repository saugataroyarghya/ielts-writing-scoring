```python
import base64, hashlib, io, json, os, pathlib, subprocess, sys, zipfile
project = pathlib.Path("/kaggle/temp/essay-scoring-code")
project.mkdir(parents=True, exist_ok=True)
payload = base64.b64decode('')
assert hashlib.sha256(payload).hexdigest() == '45f6f1abb611e362aab12d14e10dc58cbe61a51b45bc309a740b7539ccb9199a'
with zipfile.ZipFile(io.BytesIO(payload)) as archive:
    archive.extractall(project)
os.environ["PYTHONPATH"] = str(project / "src")
os.environ["TOKENIZERS_PARALLELISM"] = "false"
os.environ["HF_HUB_DISABLE_TELEMETRY"] = "1"
os.environ["HF_HOME"] = "/kaggle/temp/huggingface"
os.environ["WANDB_DISABLED"] = "true"
os.chdir(project)
print("Source bundle SHA256:", '45f6f1abb611e362aab12d14e10dc58cbe61a51b45bc309a740b7539ccb9199a')
pathlib.Path("/kaggle/working/source_bundle_sha256.txt").write_text('45f6f1abb611e362aab12d14e10dc58cbe61a51b45bc309a740b7539ccb9199a')
```

```python
dependencies = "/kaggle/temp/essay-scoring-dependencies"
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "--no-deps", "--target", dependencies, "transformers==4.57.6", "huggingface-hub==0.36.0", "tokenizers==0.22.2", "safetensors==0.7.0"], check=True)
os.environ["PYTHONPATH"] = dependencies + os.pathsep + os.environ["PYTHONPATH"]
import torch
gpu_count = torch.cuda.device_count()
if gpu_count == 0:
    raise RuntimeError("Kaggle did not allocate a GPU.")
processes = []
for index, head_name in enumerate(['mlp', 'moe']):
    command = [sys.executable, "-m", "ielts_scoring.train", "--track", 'ielts', "--head", head_name, "--mode", "train", "--data-root", "/kaggle/input", "--output", "/kaggle/working/" + head_name, "--source-hash", '45f6f1abb611e362aab12d14e10dc58cbe61a51b45bc309a740b7539ccb9199a']
    worker_env = dict(os.environ, CUDA_VISIBLE_DEVICES=str(index % gpu_count))
    if gpu_count > 1:
        processes.append((head_name, subprocess.Popen(command, env=worker_env)))
    else:
        subprocess.run(command, env=worker_env, check=True)
failures = [(name, process.wait()) for name, process in processes]
if any(code != 0 for _, code in failures):
    raise RuntimeError("A training run failed: " + repr(failures))
subprocess.run([sys.executable, "-m", "ielts_scoring.finalize", "--track", 'ielts', "--runs", "/kaggle/working"], check=True)
```
